```python
N = 4

def calculate_cost(board):
    cost = 0

    for i in range(N):
        for j in range(i + 1, N):
            if abs(board[i] - board[j]) == abs(i - j):
                cost += 1

    return cost


def display_board(board):
    for row in range(N):
        for col in range(N):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()

    print("Cost:", calculate_cost(board))


def get_neighbors(board):
    neighbors = []
    for i in range(N):
        for j in range(i + 1, N):
            new_board = board.copy()

            new_board[i], new_board[j] = new_board[j], new_board[i]

            neighbors.append(((i, j), new_board))

    return neighbors


def hill_climbing():
    current = [3, 1, 2, 0]

    step = 1

    while True:

        current_cost = calculate_cost(current)

        print("\n========== STEP", step, "==========")

        print("Current State:", current)
        print("Current Cost:", current_cost)

        display_board(current)
        if current_cost == 0:
            print("\nGoal State Reached!")
            return current

        neighbors = get_neighbors(current)

        print("\nNeighbor States and Costs:")

        best_neighbor = None
        best_cost = float('inf')
        best_pair = None

        for pair, neighbor in neighbors:

            neighbor_cost = calculate_cost(neighbor)

            print("Swap", pair,
                  ":", neighbor,
                  " Cost =", neighbor_cost)

            if neighbor_cost < best_cost:
                best_cost = neighbor_cost
                best_neighbor = neighbor
                best_pair = pair

        print("\nBest Neighbor:", best_neighbor)
        print("Best Swap:", best_pair)
        print("Neighbor Cost:", best_cost)
        if best_cost >= current_cost:
            print("\nLocal Minimum Reached!")
            return current

        current = best_neighbor
        step += 1

solution = hill_climbing()

print("\n========== FINAL SOLUTION ==========")

print("Final State:", solution)

display_board(solution)

print("\n1WN24CS064")
print("Bhuvan")



```
